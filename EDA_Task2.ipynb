# Task 2: Exploratory Data Analysis (EDA)

**Dataset:** Iris

This notebook covers descriptive statistics, histograms, boxplots, correlation analysis, pairplot, pattern/anomaly review, and feature-level inferences.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

df = pd.read_csv("data/iris.csv")
df.head()
```

```
print("Shape:", df.shape)
print("\nData types:")
print(df.dtypes)
print("\nMissing values:")
print(df.isnull().sum())
```

```
numeric_cols = df.select_dtypes(include=np.number).columns
summary = df[numeric_cols].describe().T
summary["median"] = df[numeric_cols].median()
summary["skewness"] = df[numeric_cols].skew()
summary
```

```
plt.figure(figsize=(9,6))
for col in numeric_cols:
    plt.hist(df[col], bins=12, alpha=0.45, label=col)
plt.title("Distribution of Numeric Features")
plt.xlabel("Value (cm)")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

```
plt.figure(figsize=(9,5))
sns.boxplot(data=df[numeric_cols])
plt.title("Boxplots of Numeric Features")
plt.show()
```

```
corr = df[numeric_cols].corr()
plt.figure(figsize=(7,5))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation Matrix")
plt.show()
```

```
sns.pairplot(df, hue="species", diag_kind="hist")
plt.show()
```

```
fig = px.scatter(df, x="petal_length_cm", y="petal_width_cm",
                 color="species", title="Petal Length vs Petal Width")
fig.show()
```

## EDA Findings

- 150 observations and four numeric measurements are available, grouped into three species.
- There are no missing values in the included dataset.
- Petal length and petal width show a strong positive relationship.
- The pairplot shows clearer species separation with petal measurements than with sepal measurements.
- Boxplots summarize spread and highlight observations at distribution extremes for further investigation.
- Histograms show the distributions and help assess concentration and skewness before modeling.
